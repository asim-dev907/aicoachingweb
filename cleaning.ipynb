```python
import pandas as pd

# Step 1: Create a copy of original dataset
cleaned_df = df.copy()

# Step 2: Empty spaces ko missing values mein convert karo
cleaned_df = cleaned_df.replace(
    r'^\s*$',
    pd.NA,
    regex=True
)

# Step 3: Check missing values before cleaning
print("Missing Values Before Cleaning:")
print(cleaned_df.isnull().sum())

# Step 4: Remove rows where question is missing
cleaned_df = cleaned_df.dropna(subset=["question"])

# Step 5: Fill missing categories
cleaned_df["category"] = cleaned_df["category"].fillna("Unknown")

# Step 6: Fill missing difficulty
cleaned_df["difficulty"] = cleaned_df["difficulty"].fillna("Unknown")

# Step 7: Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(cleaned_df.isnull().sum())

# Step 8: Compare dataset shape
print("\nOriginal Dataset Shape:", df.shape)
print("After Missing Values Cleaning:", cleaned_df.shape)
```

Text Cleaning (NLP Preprocessing)

```python
import re

# Original question ko preserve karna
cleaned_df["question_clean"] = cleaned_df["question"].astype(str)

# 1. Convert text into lowercase
cleaned_df["question_clean"] = cleaned_df["question_clean"].str.lower()

# 2. Remove HTML tags
cleaned_df["question_clean"] = cleaned_df["question_clean"].str.replace(
    r"<.*?>", "", regex=True
)

# 3. Remove URLs
cleaned_df["question_clean"] = cleaned_df["question_clean"].str.replace(
    r"http\S+|www\S+", "", regex=True
)

# 4. Remove punctuation and special characters
cleaned_df["question_clean"] = cleaned_df["question_clean"].str.replace(
    r"[^a-zA-Z0-9\s]", "", regex=True
)

# 5. Remove extra spaces
cleaned_df["question_clean"] = cleaned_df["question_clean"].str.replace(
    r"\s+", " ", regex=True
).str.strip()

# Display comparison
print("Original Questions:")
print(cleaned_df["question"].head(10))

print("\nCleaned Questions:")
print(cleaned_df["question_clean"].head(10))
```

Data Standardization

```python
# STEP 6: DATA STANDARDIZATION

# 1. Clean category column
cleaned_df["category"] = (
    cleaned_df["category"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# 2. Standardize category names
cleaned_df["category"] = cleaned_df["category"].str.title()


# 3. Clean difficulty column
cleaned_df["difficulty"] = (
    cleaned_df["difficulty"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# 4. Standardize difficulty values
cleaned_df["difficulty"] = cleaned_df["difficulty"].replace({
    "easy": "Easy",
    "medium": "Medium",
    "hard": "Hard"
})


# 5. Display standardized categories
print("Standardized Categories:")
print(cleaned_df["category"].value_counts())


# 6. Display standardized difficulty
print("\nStandardized Difficulty:")
print(cleaned_df["difficulty"].value_counts())


# 7. Check unique values
print("\nUnique Categories:")
print(cleaned_df["category"].unique())

print("\nUnique Difficulties:")
print(cleaned_df["difficulty"].unique())
```
